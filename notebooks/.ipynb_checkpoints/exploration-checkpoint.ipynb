# Energy USA Data Exploration

This notebook provides a template for connecting to the local database and exploring the energy data.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine
from energyusa.config import Config

# Connect to the database
# Ensure your .env file is loaded or environment variables are set
engine = create_engine(Config.DATABASE_URL)
print(f"Connected to: {Config.DATABASE_URL}")
```

## Production Data Analysis

Query aggregated production data by year and fuel type.

```python
# Query Production table
query = """
SELECT 
    date_trunc('year', date) as year, 
    fuel_type, 
    sum(value) as total_value
FROM analysis.production
GROUP BY 1, 2
ORDER BY 1, 2
"""

df_prod = pd.read_sql(query, engine)
df_prod.head()
```

```python
if not df_prod.empty:
    plt.figure(figsize=(12, 6))
    sns.lineplot(data=df_prod, x='year', y='total_value', hue='fuel_type', marker='o')
    plt.title('Annual Energy Production by Fuel Type')
    plt.ylabel('Energy Produced (Units vary)')
    plt.xlabel('Year')
    plt.legend(title='Fuel Type')
    plt.grid(True)
    plt.show()
else:
    print("No production data found. Run 'energyusa extract ...' and 'energyusa analyze' first.")
```

## Consumption Data Analysis

```python
query_cons = """
SELECT 
    date_trunc('year', date) as year, 
    sector, 
    sum(value) as total_consumption
FROM analysis.consumption
GROUP BY 1, 2
ORDER BY 1, 2
"""

df_cons = pd.read_sql(query_cons, engine)
df_cons.head()
```

```python
if not df_cons.empty:
    plt.figure(figsize=(12, 6))
    sns.barplot(data=df_cons, x='year', y='total_consumption', hue='sector')
    plt.title('Annual Energy Consumption by Sector')
    plt.ylabel('Consumption (Units vary)')
    plt.xlabel('Year')
    plt.xticks(rotation=45)
    plt.show()
else:
    print("No consumption data found.")
```
